## **CSI(Composite Share Issuance): 주식수 변화율 팩터**

**전략 가정**: 피자(기업) 크기가 그대로여도 조각(주식)을 더 내서 새 사람에게 나눠주면 내가 가진 한 조각의 몫은 작아지고, 반대로 조각 수가 줄면 한 조각의 몫은 커진다.
그래서 주식수 감소는 주주가치 제고, 주식수 증가는 주주가치 희석일 것이라는 가정에서 출발한다.

- 주식수가 **줄어드는** 기업 → 한 주의 몫이 커짐 → **롱**
- 주식수가 **늘어나는** 기업 → 한 주의 몫이 작아짐 → **숏**

시가총액 변화에서 주가 변화분을 빼면 순수한 발행주식수 변화만 남는다(아래 팩터 도출).

참고: Daniel & Titman (2006, *Journal of Finance*), Pontiff & Woodgate (2008, *Journal of Finance*)

---
**팩터 도출**

시가총액은 주가와 발행주식수의 곱이다.

$$ME_t = P_t \times N_t \quad\Longrightarrow\quad \log ME_t = \log P_t + \log N_t$$

$t-\tau$ 시점과 $t$ 시점을 비교하면:

$$\log\frac{ME_t}{ME_{t-\tau}} = \underbrace{[\log P_t - \log P_{t-\tau}]}_{\text{주가 변화분}} + \underbrace{[\log N_t - \log N_{t-\tau}]}_{\text{주식수 변화분}}$$

여기서 주가수익률 $r(t-\tau,t) = \log P_t - \log P_{t-\tau}$(배당 미반영, 분할·무상증자만 조정)를 빼면 주가 변화분이 상쇄되고 순수한 주식수 변화율만 남는다:

$$\log\frac{ME_t}{ME_{t-\tau}} - r(t-\tau,t) = \log\frac{N_t}{N_{t-\tau}}$$

필요 데이터는 2개뿐이다: **시가총액**(월말), **배당 미반영 수정주가**(월말 — 주가수익률 계산용)

---
**왜 배당 미반영 수정주가를 써야 하나**

- 무상증자·액면분할은 피자를 더 잘게 써는 것뿐, 피자 크기(기업가치) 자체는 그대로다. 원본 주가를 쓰면 "주식 수 2배 → 숏 시그널"과 "주가 반토막 → 숏 대박"이라는 착시 손익이 동시에 발생한다.
- 배당 미반영 수정주가는 과거 주가도 같은 비율로 소급 조정해서, 이런 이벤트의 주가 변동률을 $0$으로 만들어 $r(t-\tau,t)$에서 자동으로 상쇄시킨다 — 가짜 시그널 차단.
- 반면 유상증자·CB전환(주식수 증가)이나 자사주 소각 등(주식수 감소)은 수정계수에 반영되지 않으므로, 배당 미반영 수정주가를 써도 시그널은 그대로 살아남는다 — 실제 주식수 변화만 롱/숏으로 포착.

**수식으로 보기 — 배당 미반영 수정주가가 왜 수익률을 $0$으로 만드는가**

무상증자·분할 비율 $k$ (주식수 $k$배, 가치불변이니 주가는 $1/k$배):

$$r_{raw}=\log P_{t_0^+}-\log P_{t_0^-}=-\log k \quad\text{(원본 주가 수익률은 이벤트에서 이만큼 점프)}$$

배당 미반영 수정주가는 과거 가격을 $1/k$로 소급 조정해 만든 것이므로, 이 점프가 정확히 상쇄되어

$$r_{adj}=-\log k+\log k=0$$

이 $-\log k$가 원본 주가를 쓰면 팩터($\iota$) 안에 그대로 남아 "가짜 발행"처럼 잡히고, 배당 미반영 수정주가를 쓰면 $0$이 되어 사라진다.

## **2. CSI(주식수 변화율) 팩터**

τ(=48)개월간 로그 시가총액 변화에서 같은 기간 누적 주가수익률(로그, 배당 미반영)을 빼면 주가 효과가 상쇄되어 순수한 주식수 변화율만 남는다.

$$\iota_i(t-\tau,\,t)=\big[\log ME_{i,t}-\log ME_{i,t-\tau}\big]-\sum_{s=t-\tau+1}^{t}\log(1+r_{i,s})$$

- iota < 0 → 주식수 감소(자사주 소각 등) → 롱 후보
- iota > 0 → 주식수 증가(유상증자·전환사채 전환 등) → 숏 후보

```python
# [CSI 팩터 핵심]
# τ개월 로그 시가총액 변화에서 같은 기간의 배당 미반영 수정주가 로그수익률을 빼 주가 효과를 제거한다.
# 남는 값은 발행주식수의 로그 변화율에 해당한다. 값이 작을수록 주식수 감소, 클수록 주식수 증가 신호다.
TAU = 48   # 관측기간(개월)
log_me         = np.log(market_cap_for_log)  # 0 이하를 제외한 월말 시가총액 로그값
cum_log_return = np.log1p(return_for_log).rolling(TAU).sum()  # -100% 이하를 제외한 배당 미반영 누적 로그 주가수익률
csi_factor = (log_me - log_me.shift(TAU)) - cum_log_return   # 시총 변화에서 주가 변화를 제거한 CSI; 낮은 값=롱, 높은 값=숏
csi_factor['삼성전자'].dropna().tail()   # 예시: 삼성전자 (자사주 소각으로 음수)
```

### **3-1. 유니버스와 종목 선정**

1. **시가총액 상위 30%**: 리밸런싱일 월말 시가총액 기준.
2. **롱**: 48개월간 실제로 주식수가 줄어든 종목(CSI < −0.5%) 가운데 CSI가 가장 낮은 30개를 담는다. 30개보다 적으면 있는 만큼만 담는다
   (2012~2018년에는 19~29개로 모자란 때가 있고, 2019년부터는 항상 30개 이상이다).
3. **숏**: CSI 상위 30종목(주식수 증가).

```python
# 3-1. 유니버스·거래 가능 종목·종목 선정
_liq_px_cache = {}
# 종목별 유효 일별 배당반영 수정주가를 한 번만 정리해 캐시에 저장하고 이후 체결·평가에서 재사용한다
def _liq_price_series(ticker):
    """종목별 유효 일별 배당반영 수정주가를 한 번만 정리해 캐시에 저장하고 이후 체결·평가에서 재사용한다."""
    # 캐시에 없는 종목만 원본 가격에서 결측치를 제거해 새로 저장한다.
    if ticker not in _liq_px_cache:
        _liq_px_cache[ticker] = adj_close[ticker].dropna()
    return _liq_px_cache[ticker]


_daily_ret     = adj_close.pct_change(fill_method=None)
_vol60_monthly = (_daily_ret.rolling(60, min_periods=40).std() * np.sqrt(252)).resample('ME').last()   # 60거래일 롤링 변동성(연환산), 월말 값


# 시총 상위 30% 유니버스에서 CSI 순위로 뽑는다
def select_candidates(date, P):
    """시총 상위 30% 유니버스에서 CSI 순위로 뽑는다.
    롱: 실제로 주식수가 줄어든(CSI < -0.5%) 종목만, CSI가 낮은 순서. 숏: CSI가 높은 순서. 두 레그 모두 앞에서 최대 n종목을 담는다."""
    cap_today = market_cap_monthly.loc[date].dropna()  # 신호일에 시가총액이 존재하는 종목만 출발점으로 사용
    universe = cap_today[cap_today >= cap_today.quantile(P['cap_pct'])].index  # 70% 분위 경계 이상, 즉 시가총액 상위 약 30%
    ff = P['factor'].loc[date].reindex(universe).dropna()  # 유니버스 안에서 CSI가 계산 가능한 종목만 순위화
    # 유동성은 여기서 순위에 반영하지 않는다. 후보·순위는 CSI로만 정하고 실제 체결 가능성은 3-4 엔진에서 처리한다.
    long_c = ff[ff < -REAL_CHANGE_EPS]                    # CSI<-0.5%: 측정오차를 넘는 실제 주식수 감소 종목만 롱 후보
    long_ranked = list(long_c.sort_values(ascending=True, kind='mergesort').index)   # CSI가 가장 낮은 종목부터 롱 우선순위
    short_ranked = list(ff.sort_values(ascending=False, kind='mergesort').index)     # CSI가 가장 높은 종목부터 숏 우선순위
    return long_ranked, short_ranked
```

### **3-2. 가중 방식**

선정된 종목은 바꾸지 않고 비중만 조정한다. 포트폴리오 하나(333.3억)의 롱·숏을 각각 아래 비중대로 나눠 담는다.

**가중 계산 방법**
1. **점수**
   * 롱: 점수 = 감소규모(−CSI) × 점진성, 점진성 = 1 − (48개월 감소분 중 가장 큰 한 달이 차지하는 비중).
     여러 달에 걸쳐 나눠 줄였으면 점진성이 1에 가깝고, 한 달에 몰아서 줄었으면(일회성 사건) 0에 가깝다.
     월 주식수 변화가 −0.1%보다 작은 달만 '감소한 달'로 센다(시가총액 반올림 때문에 생기는 ±0.001% 수준의 노이즈 제외). 감소한 달이 한 달뿐이면 점진성은 0이다.
   * 숏: 점수 = −(최근 60거래일 변동성). 변동성이 낮을수록 점수가 높다.
2. **순위**: 점수 순위를 0(꼴찌) ~ 1(1등)로 펼친다. 점수가 같은 종목(예: 감소한 달이 한 달뿐이라 점진성 0인 종목들)은 그 순위들의 평균을 같이 받아 같은 비중이 된다.
3. **가중** = 1 + k × (2 × 순위 − 1). 꼴찌는 1 − k, 중간은 1, 1등은 1 + k다. 롱 k=1.0이면 0~2배, 숏 k=0.5면 0.5~1.5배.
4. **목표비중**: 가중의 합이 1이 되게 나눈다.

점수 값을 그대로 쓰지 않고 순위로 바꾸는 이유: 점수 값을 그대로 쓰면 CSI나 변동성의 극단값 한두 종목에 비중이 쏠린다. 순위로 바꾸면 종목별 비중이 항상 정해진 범위(0~2배, 0.5~1.5배) 안에 머문다.

**k값이 과적합이 아닌 근거**
* **표본내에서만 판단했다**: k는 표본내(2012.03~2020.12) 성과로만 판단했다. 표본외(2021.01~2026.08)는 k를 정하는 데 쓰지 않았고, 6-2의 표본외 성과는 이렇게 고정한 k로 확인한 결과다.
* **롱 k=1.0은 신호가 나온 종목이 모두 비중을 받는 최대값이다**: 가중 = 1 + k(2·순위 − 1)에서 k=1.0이면 최하위 종목도 0배(하한 0.02) 이상이라, 실제로 주식수가 줄어든 종목이 모두 포트폴리오에 담긴다(하한에 걸리는 종목 0%).
  k를 더 키우면 순위가 낮은 종목이 하한에 걸려 사실상 빠지기 시작한다: 하한에 걸리는 롱 종목 비율이 k=2.0 11%, k=2.5 22%, k=3.0 32%, k=5.0 52%로 늘고,
  감소한 달이 한 번뿐인 종목(표본내 롱 후보의 약 70%)의 비중도 동일가중 대비 0.69배(k=1.0)에서 0.38배(k=2.0), 0.17배(k=3.0)로 줄어든다.
  "주식수가 줄어든 종목을 담는다"는 전략 취지를 지키기 위해 k=1.0으로 정했다. 표본내 Sharpe는 k=0 0.54, k=1.0 0.71, k=2.0 0.83으로 k=1.0에서도 가중 효과가 이미 나타나며(k=3.0 0.83 이후 k=5.0은 0.70으로 하락), 이 선택으로 잃는 성과도 거의 없다.
* **숏은 표본내 Sharpe 최대값을 오히려 피했다**: 표본내 Sharpe만 보면 가중이 없는 쪽(k=0)이 가장 높지만, 표본내 MDD도 k=0이 가장 깊다. 변동성이 큰 종목을 숏하면 숏스퀴즈형 급등에 노출되는데,
  Sharpe는 표준편차만 벌점을 줄 뿐 특정 시점에 몰린 깊은 낙폭은 따로 잡지 못한다. Sharpe와 MDD를 함께 보고 k=0.5를 골랐다.
* **k를 바꿔도 성과가 완만하게 변한다**: 특정 k값에서만 성과가 튀면 과적합을 의심해야 하지만, 롱·숏 모두 정상 범위(롱 0~2, 숏 0~1) 안에서는 k를 조금씩 바꿔도 표본내 Sharpe·MDD가 완만하게 변한다.
* **가중 없이도 전략이 작동한다**: 롱·숏 모두 동일가중으로 해도 CSI 전략은 수익을 낸다(강건성 노트북 4장 2). 가중은 성과의 원천이 아니라 보조 조정이다.
* **가중 아이디어는 경제적 가설에서 출발했다**: 롱의 점진성은 "여러 달에 걸쳐 꾸준히 주식수를 줄인 기업이 한 번에 줄인 기업보다 지속적인 주주가치 제고 의지를 보인다"는 가설, 숏의 저변동성은 "고변동성 종목 숏의 스퀴즈 위험"이라는 가설이다.

**숏에 저변동성 가중을 쓰게 된 과정 (표본내)**
1. 처음에는 롱과 대칭으로, 숏 후보에도 증가규모 × 점진성 가중을 적용해 봤다("여러 달에 걸쳐 꾸준히 주식을 찍어낸 기업일수록 더 크게 숏").
2. 표본내에서 확인해보니 이 가중은 Sharpe가 더 높았다(k=1.0 기준 0.76 대 저변동성 가중 0.65). 문제는 낙폭이었다: 표본내 MDD가 −33%로 깊어졌고(저변동성 가중은 −19%),
   전략의 가장 나쁜 달도 −26%(2019년 10월)로 저변동성 가중의 가장 나쁜 달(−10%)보다 훨씬 컸다.
   가중을 크게 준 종목들은 변동성이 평균보다 높은 쪽(가중평균 60일 변동성 56% 대 동일가중 54%)에 조금 더 실려 있었다.
3. Sharpe만 보면 증가규모×점진성 가중이 더 높지만, 그 대가로 MDD가 크게 깊어지는 트레이드오프를 받아들이지 않기로 하고, 신호 강도가 아니라 보유 위험을 기준으로 비중을 나누는 저변동성 가중으로 바꿨다.
   종목 선정(CSI 상위 30)은 그대로 두고 비중만 조정한다.

k값별 표본내·전체 성과와 동일가중 비교는 `Robustness Test_last.ipynb` 4장 2)·3)에 있다.

```python
# 3-2. 가중 방식 (선정된 30종목 안에서 비중만 조정)
_d_monthly = np.log(market_cap_for_log).diff() - np.log1p(return_for_log)   # 로그 불능 관측치를 제외한 월별 주식수 로그변화


# score의 순위를 0~1로 펼친 뒤(꼴찌 0, 1등 1) 1 + k(2*순위 - 1) 가중으로 바꾼다 (k=1: 0~2배, k=0.5: 0.5~1.5배)
def _rank_tilt(score, k):
    """score의 순위를 0~1로 펼친 뒤(꼴찌 0, 1등 1) 1 + k(2*순위 - 1) 가중으로 바꾼다 (k=1: 0~2배, k=0.5: 0.5~1.5배).
    점수가 같은 종목들은 그 순위들의 평균 순위를 같이 받는다(예: 꼴찌 10개가 동점이면 1~10위의 평균).
    score가 없는(데이터 부족) 종목은 중립 가중(1.0)을 준다 — 실제 체결 여부는 3-4 체결 단계에서 결정된다."""
    s = score.dropna()
    # 종목이 하나뿐이면 순위를 정할 수 없으므로 중립 순위 0.5를 부여한다.
    pr = (s.rank(method='average') - 1) / (len(s) - 1) if len(s) > 1 else s * 0 + 0.5
    return (1 + k * (2 * pr - 1)).clip(lower=0.02).reindex(score.index).fillna(1.0)


# 관측기간(τ) 감소분 중 가장 큰 한 달이 차지하는 비중 (1 = 한 달에 몰아서 감소, 0에 가까울수록 여러 달에 나눠 감소)
def _lump_dn(date, tickers, tau):
    """관측기간(τ) 감소분 중 가장 큰 한 달이 차지하는 비중 (1 = 한 달에 몰아서 감소, 0에 가까울수록 여러 달에 나눠 감소).
    월 주식수 변화가 -MONTH_EPS(-0.1%)보다 작은 달만 '감소한 달'로 센다. 시가총액 반올림 때문에 생기는 ±0.001% 수준의 노이즈를 걸러낸다."""
    loc = _d_monthly.index.get_loc(date)
    w = _d_monthly.iloc[loc - tau + 1: loc + 1].reindex(columns=tickers)
    neg = (-w).where(w < -MONTH_EPS, 0.0)
    return (neg.max() / neg.sum().replace(0, np.nan)).fillna(1.0)


# 롱 목표비중(합 1). 'strategy': 감소규모 x 점진성 순위로 가중(k=long_k)
def _long_weight_target(date, top_n, P):
    """롱 목표비중(합 1). 'strategy': 감소규모 x 점진성 순위로 가중(k=long_k)."""
    # 동일가중 비교에서는 점수 계산을 건너뛰고 선택 종목에 같은 비중을 준다.
    if P['weight_mode'] == 'equal':
        return {t: 1.0 / len(top_n) for t in top_n}
    # 시가총액가중 비교에서는 신호 점수 대신 리밸런싱일 시가총액 비율을 사용한다.
    if P['weight_mode'] == 'cap':
        cap = market_cap_monthly.loc[date, top_n]
        return (cap / cap.sum()).to_dict()
    c = P['factor'].loc[date]                              # 리밸런싱일의 종목별 CSI
    real = [t for t in top_n if c[t] < -REAL_CHANGE_EPS]  # 실제 감소 기준을 다시 확인해 가중 대상 확정
    w = pd.Series(1.0, index=top_n)                        # 데이터 부족 시 사용할 중립 가중의 출발값
    # 순위를 계산할 실제 감소 종목이 2개 이상일 때만 점진성 가중을 적용하고, 0~1개이면 중립 가중을 유지한다.
    if len(real) >= 2:
        score = (-c.reindex(real)) * (1 - _lump_dn(date, real, P['tau']))  # 감소규모 × 점진성
        tw = _rank_tilt(score, P['long_k'])  # 점수의 절대크기 대신 순위를 사용해 극단치 집중을 완화
        w.loc[real] = tw / tw.sum() * len(real)
    return (w / w.sum()).to_dict()


# 숏 목표비중(합 1). 'strategy': 60거래일 변동성이 낮을수록 가중(k=short_k). 고변동성(숏스퀴즈 위험) 종목의 비중을 줄인다
def _short_weight_target(date, top_n, P):
    """숏 목표비중(합 1). 'strategy': 60거래일 변동성이 낮을수록 가중(k=short_k). 고변동성(숏스퀴즈 위험) 종목의 비중을 줄인다."""
    # 동일가중 비교에서는 점수 계산을 건너뛰고 선택 종목에 같은 비중을 준다.
    if P['weight_mode'] == 'equal':
        return {t: 1.0 / len(top_n) for t in top_n}
    # 시가총액가중 비교에서는 신호 점수 대신 리밸런싱일 시가총액 비율을 사용한다.
    if P['weight_mode'] == 'cap':
        cap = market_cap_monthly.loc[date, top_n]
        return (cap / cap.sum()).to_dict()
    v = _vol60_monthly.loc[date].reindex(top_n)  # 최근 60거래일 연환산 변동성
    w = _rank_tilt(-v, P['short_k'])              # 변동성이 낮을수록 높은 점수: 숏스퀴즈 위험 완화 목적
    return (w / w.sum()).to_dict()
```

### **3-4. 거래대금 제약 없는 체결**

Robustness Test 노트북의 `participation=None` 조건과 동일하게 거래대금에 따른 일별 체결 한도를 적용하지 않는다.

* **편입**: 리밸런싱 신호일에 가격이 있으면 그날 목표비중 전량을 체결한다.
* **정리**: 후보에서 빠졌거나 목표비중을 초과한 종목도 가격이 있으면 그날 전량 정리한다.
* **미체결**: 가격이 없는 종목은 체결하지 않는다. 거래대금 부족에 따른 부분 체결·현금 잔여는 발생하지 않는다.

체결가격은 기존 엔진과 같이 배당반영 수정주가이며, 호가충격과 장중 슬리피지는 별도로 반영하지 않는다.


```python
# 3-4. 거래대금 제약 없는 체결 (레그 하나, 리밸런싱 구간 하나)
# 상장폐지 등으로 리밸런싱일 이후 가격이 없는 종목이면 False (잔여 포지션은 마지막 가격에 청산된 것으로 본다)
# [상장폐지 종목 처리 점검: 수정하지 않는 근거]
# _liq_alive는 전체 데이터에서 종목의 마지막 가격일(px.index[-1])을 보므로, 형식상 리밸런싱일(day0) 이후 정보를 쓴다.
# 다만 실제로 판정에 쓰이는 정보는 day0 시점에 이미 공개된 사실이라, 성과에 미래 정보가 들어가지 않는다고 판단해 그대로 둔다.
#
# 1) False가 나온 경우: 전체 백테스트에서 42회(15종목). 모두 마지막 거래일이 day0보다 7~86일 앞선다.
#    → day0에는 이미 상장폐지가 끝나 공개된 상태였다(거래정지 중 재개 여부를 미리 안 경우는 없음).
#
# 2) 종목 유형:
#    - 합병·공개매수 후 자진 상장폐지(우리금융지주, 현대하이스코, 오스템임플란트, 락앤락, 커넥트웨이브 등 약 10종목):
#      주주는 합병 신주나 공개매수 대금으로 대략 마지막 가격만큼 받으므로, 마지막 가격 청산이 현실에 가깝다.
#    - 부실 상장폐지(쌍용건설, 이아이디 등 5종목): 정리매매 폭락(마지막 20거래일 -83~-99%)이 이미 가격 데이터에 들어 있어
#      손실이 반영된 뒤이고, 남은 비중도 포트폴리오의 0.01~0.1% 수준이다.
#
# 3) 극단 가정으로 다시 돌린 결과 (거래대금 제약 없음, 표본외 2021.01~2026.08):
#    - 현재 코드(마지막 가격 청산):                         표본외 Sharpe 1.87, CAGR 27.3%
#    - 미래 정보 없이 영원히 보유(마지막 가격 고정, 자금 묶임):   표본외 Sharpe 1.76, CAGR 23.3%
#    - 폐지된 롱 종목 -100% 손실:                           표본외 Sharpe 1.72, CAGR 25.6%
#    두 대안 모두 현실보다 가혹한 하한이고, 이 경우에도 표본외 Sharpe는 1.7대로 결론이 바뀌지 않는다.
#
# 4) 거래정지 후 재개된 종목은 True로 계속 보유(정지 기간 매도 불가)되어 현실과 같다.
# 월말 평가에서 한 구간 내내 가격이 없는 보유 종목을 빠뜨리는지도 따로 확인했다: 보정해도 결과 동일(해당 사례 0건).
# 엄밀하게 하려면 KRX 상장폐지일(당시 공시 정보)로 판정해야 하지만, 위 결과상 성과 수치에는 영향이 없다고 본다.
def _liq_alive(ticker, day0):
    """상장폐지 등으로 리밸런싱일 이후 가격이 없는 종목이면 False (잔여 포지션은 마지막 가격에 청산된 것으로 본다)."""
    px = _liq_price_series(ticker)
    return len(px) > 0 and px.index[-1] >= day0


# 구간 안의 월말 평가 시점 (마지막은 구간 종료일)
def _checkpoints(start_date, end_date, calendar):
    """구간 안의 월말 평가 시점 (마지막은 구간 종료일)."""
    cps = [d for d in calendar if start_date < d <= end_date]
    # 월말 체크포인트가 없거나 마지막 평가일이 빠졌으면 구간 종료일을 최종 평가시점으로 추가한다.
    if not cps or cps[-1] != end_date:
        cps = cps + [end_date]
    return cps


# 레그(롱 또는 숏) 하나를 한 리밸런싱 구간 동안 하루 단위로 체결한다
def _liq_leg_step_ckpt(prev_w, ranked_candidates, start_date, end_date, n, target_total, top_targets, checkpoint_dates,
                       aum, participation, max_days):
    """레그(롱 또는 숏) 하나를 한 리밸런싱 구간 동안 하루 단위로 체결한다.
    - 종목당 하루 한도 = 그날 기준 20일 평균거래대금 x participation / aum (매매일마다 재계산)
      (participation=None이면 한도 없이 가격이 있는 날 전량 체결 — 강건성 노트북의 거래대금 제약 유무 비교에서만 쓴다)
    - 편입: 리밸런싱일부터 max_days 영업일 동안 상위 n종목을 목표비중까지 매수. 못 채운 몫은 현금(후순위로 넘기지 않음)
    - 정리: 후보에서 빠진 종전 보유 종목은 구간 끝까지 같은 한도로 매도. 못 판 종목만 다음 구간으로 이월
    반환: (기여수익률, 다음 구간 시작 비중, 거래 비중 합, 보유 비중 합, 편입 후 종목별 비중,
          월말 평가 값, 월말까지 누적 거래 비중)"""
    D = max_days
    base = target_total / n
    top_targets = top_targets or {}
    # 종목 t의 목표 명목비중을 반환한다. 별도 가중이 없으면 레그 자금을 n등분한다
    def tgt(t):
        """종목 t의 목표 명목비중을 반환한다. 별도 가중이 없으면 레그 자금을 n등분한다."""
        return top_targets.get(t, base)
    cal = adj_close.index
    pre = cal[cal <= start_date]
    # 캘린더에 기준일 이전 거래일이 없을 때만 입력 시작일을 예외적인 대체값으로 사용한다.
    day0 = pre[-1] if len(pre) else start_date
    days = cal[(cal >= day0) & (cal <= end_date)]
    ranked = list(ranked_candidates)

    hold = {t: float(w) for t, w in prev_w.items() if abs(w) > 1e-12 and _liq_alive(t, day0)}
    w0 = dict(hold)
    buys, sells = {}, {}
    px_d, cap_d = {}, {}

    # 체결 시뮬레이션에 필요한 종목별 가격 배열과 일별 주문 한도 배열을 준비한다
    def prep(t):
        """체결 시뮬레이션에 필요한 종목별 가격 배열과 일별 주문 한도 배열을 준비한다."""
        # 같은 종목의 가격·한도 배열을 반복 생성하지 않도록 최초 접근 때만 준비한다.
        if t not in px_d:
            px_d[t] = _liq_price_series(t).reindex(days).values
            # 거래대금 제약을 끈 실행에서는 일별 주문 한도를 무한대로 두어 유효가격이 있으면 즉시 목표량을 체결한다.
            if participation is None:
                cap_d[t] = np.full(len(days), np.inf)
                return
            # 거래대금 자료가 있는 종목만 ADV 기반 한도를 계산하고, 자료가 없으면 한도를 0으로 두어 체결하지 않는다.
            if t in trading_value_20.columns:
                adv_arr = trading_value_20[t].reindex(days).values
            else:
                adv_arr = np.full(len(days), np.nan)
            cap_d[t] = np.where(pd.notna(adv_arr) & (adv_arr > 0), participation * adv_arr / aum, 0.0)

    for t in list(hold) + ranked[:n]:
        prep(t)

    # k번째 거래일의 유효한 양(+)의 배당반영 수정주가를 반환하고, 거래 불가능하면 None을 반환한다
    def price(t, k):
        """k번째 거래일의 유효한 양(+)의 배당반영 수정주가를 반환하고, 거래 불가능하면 None을 반환한다."""
        p = px_d[t][k]
        return p if (p == p and p > 0) else None

    group, selected = ranked[:n], set(ranked[:n])
    # cash는 실제 원화계좌 잔액이 아니라 고정 AUM=1 안에서 아직 종목에 배분하지 않은 명목비중이다.
    # 이전 구간의 손익으로 다음 주문 원금을 늘리거나 줄이지 않으며, 매도손익도 추가 투자원금으로 재투자하지 않는다.
    cash = target_total - sum(hold.values())

    # 청산 제한을 두지 않으면 구간 끝까지 매도하고, 제한하는 경우에는 편입기간 D까지만 처리한다.
    k1 = len(days) if EXIT_UNLIMITED else min(D, len(days))
    for k in range(0, k1):
        room = {}
        for t in set(group) | set(hold):
            prep(t)
            room[t] = cap_d[t][k] if price(t, k) is not None else 0.0
        for t in list(hold):                                # 매도: 후보에서 빠진 종목 / 목표 초과분
            keep = tgt(t) if t in selected else 0.0
            # 편입기간이 지난 뒤에도 후보로 남은 종목은 추가 비중조정을 멈추고 현재 보유량을 유지한다.
            if k >= D and t in selected:
                continue
            over = hold[t] - keep
            sold_so_far = sum(x for x, _, _ in sells.get(t, []))
            left_w0 = w0.get(t, 0.0) - sold_so_far
            # 목표 초과분, 당일 체결여력, 기존 보유량이 모두 있을 때만 매도수량을 계산한다.
            if over > 1e-12 and room.get(t, 0.0) > 1e-12 and left_w0 > 1e-12:
                sq = min(over, room[t], left_w0)
                # 부동소수점 오차 수준의 미세 매도는 거래로 기록하지 않는다.
                if sq > 1e-12:
                    sells.setdefault(t, []).append((sq, price(t, k), k))
                    hold[t] -= sq; room[t] -= sq
                    cash += sq  # 실제 매도대금이 아니라 청산된 명목비중만 고정 AUM 내 매수 여력으로 복원
        # 신규·추가 매수는 편입기간 D 안에서만 허용하고 이후에는 빈 목록으로 처리한다.
        for t in (group if k < D else []):                   # 매수: 편입 창(D영업일) 안에서만
            need = max(tgt(t) - hold.get(t, 0.0), 0.0)
            q = min(room.get(t, 0.0), need, max(cash, 0.0))
            # 부동소수점 오차 수준의 미세 매수는 거래로 기록하지 않는다.
            if q > 1e-12:
                buys.setdefault(t, []).append((q, price(t, k), k))
                hold[t] = hold.get(t, 0.0) + q; room[t] -= q; cash -= q

    contrib_total, traded_total, achieved_total = 0.0, 0.0, 0.0
    end_vals, p_start_map = {}, {}
    for t in set(hold) | set(buys) | set(sells):
        px = _liq_price_series(t)
        before = px.loc[:start_date]
        during = px[(px.index > start_date) & (px.index <= end_date)]
        has_start_price = len(before) > 0 and pd.notna(before.iloc[-1]) and before.iloc[-1] > 0
        has_end_price = len(during) > 0 and pd.notna(during.iloc[-1]) and during.iloc[-1] > 0
        p_start_map[t] = before.iloc[-1] if has_start_price else np.nan
        sold = sum(x for x, _, _ in sells.get(t, []))
        keep = w0.get(t, 0.0) - sold
        # 구간 안에 시작·종료 평가가격이 모두 있을 때만 실제 가격수익률을 반영한다.
        if has_start_price and has_end_price:
            p_start, p_end = before.iloc[-1], during.iloc[-1]
            held_ret = min(p_end / p_start - 1, WINSORIZE_CAP)
            contrib = keep * held_ret
            contrib += sum(x * min(p_sell / p_start - 1, WINSORIZE_CAP) for x, p_sell, _ in sells.get(t, []))
            val = keep * (1 + held_ret)
            for x, p_buy, _ in buys.get(t, []):
                buy_ret = min(p_end / p_buy - 1, WINSORIZE_CAP)
                contrib += x * buy_ret
                val += x * (1 + buy_ret)
        else:
            # 구간 중 새로운 유효가격이 없으면 해당 종목의 수익률을 0%로 처리한다.
            contrib, val = 0.0, keep + sum(x for x, _, _ in buys.get(t, []))
        contrib_total += contrib
        end_vals[t] = val
        traded_total += sold + sum(x for x, _, _ in buys.get(t, []))
        achieved_total += keep + sum(x for x, _, _ in buys.get(t, []))

    end_series = pd.Series(end_vals, dtype=float)
    next_w = end_series[end_series.abs() > 1e-12]

    # 월말 평가: 그 시점까지 실제 체결된 매수·매도와 그날 배당반영 수정주가만 사용(보간 없음)
    ckpt_vals, ckpt_traded = {}, {}
    days_vals = days.values
    for cp in checkpoint_dates:
        k_cp = np.searchsorted(days_vals, np.datetime64(cp), side='right') - 1
        # 체크포인트 이전에 사용할 거래일이 하나도 없으면 해당 평가시점을 건너뛴다.
        if k_cp < 0:
            continue
        ckpt_traded[cp] = (sum(x for lst in buys.values() for x, _, kk in lst if kk <= k_cp)
                           + sum(x for lst in sells.values() for x, _, kk in lst if kk <= k_cp))
        cash_now = target_total - sum(w0.values())
        for t, lst in buys.items():
            cash_now -= sum(x for x, _, kk in lst if kk <= k_cp)
        for t in set(w0) | set(sells):
            p_start = p_start_map.get(t, np.nan)
            # 시작가격이 NaN이거나 0 이하이면 수익률 계산이 불가능하므로 해당 종목의 평가를 건너뛴다.
            if not (p_start == p_start and p_start > 0):
                continue
            for sq, p_sell, kk in sells.get(t, []):       # 매도 대금은 매도 시점까지의 수익을 반영한 금액으로 현금화
                # 체크포인트까지 실제로 발생했고 매도가격도 유효한 거래만 현금화 금액에 반영한다.
                if kk <= k_cp and p_sell == p_sell and p_sell > 0:
                    cash_now += sq * (1 + min(p_sell / p_start - 1, WINSORIZE_CAP))
        val = cash_now
        for t in set(w0) | set(buys) | set(sells):
            p_start = p_start_map.get(t, np.nan)
            # 시작가격이 NaN이거나 0 이하이면 수익률 계산이 불가능하므로 해당 종목의 평가를 건너뛴다.
            if not (p_start == p_start and p_start > 0):
                continue
            keep_now = w0.get(t, 0.0) - sum(x for x, _, kk in sells.get(t, []) if kk <= k_cp)
            p_cp = px_d[t][k_cp] if t in px_d else np.nan
            # 체크포인트 당일 가격이 없으면 그 시점 이전의 마지막 유효가격을 찾는다.
            if not (p_cp == p_cp and p_cp > 0):
                arr = px_d.get(t)
                # 가격 배열이 준비된 종목에 한해서 과거 유효가격을 탐색한다.
                if arr is not None:
                    valid_idx = np.where(pd.notna(arr[:k_cp + 1]) & (arr[:k_cp + 1] > 0))[0]
                    p_cp = arr[valid_idx[-1]] if len(valid_idx) else np.nan
            # 아직 보유 중인 수량과 유효 평가가격이 있을 때만 보유가치를 합산한다.
            if keep_now > 1e-12 and p_cp == p_cp and p_cp > 0:
                val += keep_now * (1 + min(p_cp / p_start - 1, WINSORIZE_CAP))
            for x, p_buy, kk in buys.get(t, []):
                # 체크포인트까지 체결된 매수 중 매수가격과 평가가격이 모두 유효한 건만 평가한다.
                if kk <= k_cp and p_buy == p_buy and p_buy > 0 and p_cp == p_cp and p_cp > 0:
                    val += x * (1 + min(p_cp / p_buy - 1, WINSORIZE_CAP))
        ckpt_vals[cp] = val

    return contrib_total, next_w, traded_total, achieved_total, {t: w for t, w in hold.items() if w > 1e-9}, ckpt_vals, ckpt_traded
```

### **3-5. 거래비용·수익률 계산**

* **거래비용**: 왕복 스프레드 50bp(매수·매도 각 25bp, 실제로 매매한 날이 속한 달에 부과) + 숏 대차수수료 연 3%(보유 기간에 비례).
* **월말 배당반영 수정주가 평가**: 가격 데이터는 일별 배당반영 수정주가를 사용하므로, 매월 마지막 거래일의 배당반영 수정주가로 보유 종목을 평가한다. 편입 중인 종목은 그때까지 실제로 산 몫만, 정리 중인 종목은 아직 못 판 몫만 평가하고,
  이미 판 몫은 판 날의 배당반영 수정주가로 현금화한 금액으로 잡는다. 월별 수익률 = (이번 달 말 평가액 − 지난달 말 평가액) / 지난달 말 평가액.
  리밸런싱은 3개월마다지만 수익률은 월 단위로 끊어 재므로, 구간 중간의 낙폭도 MDD에 잡힌다.
* **종목 수익률 상한**: 한 구간 수익률이 +2000%를 넘으면 데이터 오류로 보고 +2000%로 자른다(아래 셀에서 실제로 걸리는 관측치를 확인).
* **합산**: 운용규모는 1000억 고정이다. 세 포트폴리오는 항상 333.3억씩이므로(수익은 인출·손실은 보전해 매달 333.3억으로 맞춘다고 가정) port_A·B·C의 월별 수익률을 평균한다.
  거래대금 제약은 적용하지 않으므로 종목별 하루 체결 한도는 계산하지 않는다.

```python
# 3-5. 구간 수익률(비용·월말 배당반영 수정주가 평가) → 포트폴리오 → port_A·B·C 합산
# 리밸런싱 구간 하나(start_date~end_date)를 계산한다. 전체 백테스트와 첫 리밸런싱 예시가 모두 이 함수를 쓴다
def strategy_window(start_date, end_date, prev_long_w, prev_short_w, cps, P):
    """리밸런싱 구간 하나(start_date~end_date)를 계산한다. 전체 백테스트와 첫 리밸런싱 예시가 모두 이 함수를 쓴다.
    월별 수익률 = (이번 체크포인트 평가액 - 직전 체크포인트 평가액) / 직전 체크포인트 평가액 (구간 시작 평가액 1.0).
    스프레드 비용은 실제로 매매한 날이 속한 달에, 대차수수료는 경과일에 비례해 부과한다."""
    n = P['n']
    long_ranked, short_ranked = select_candidates(start_date, P)
    lt = _long_weight_target(start_date, long_ranked[:n], P)
    st = _short_weight_target(start_date, short_ranked[:n], P)
    L = _liq_leg_step_ckpt(prev_long_w, long_ranked, start_date, end_date, n, 1.0, lt, cps, P['aum'], P['participation'], P['max_days'])
    S = _liq_leg_step_ckpt(prev_short_w, short_ranked, start_date, end_date, n, 1.0, st, cps, P['aum'], P['participation'], P['max_days'])
    rL, next_long_w, traded_l, achieved_l, hold_l, ckptL, cumL = L
    rS, next_short_w, traded_s, achieved_s, hold_s, ckptS, cumS = S

    rets, costs = {}, {}
    long_vals, short_vals = {}, {}   # 레그별 기여 분해용: 체크포인트별 레그 자체 가치(구간 시작=1.0, 비용 반영 전)
    prev_cpL = prev_cpS = V = 1.0
    prev_cp_date, prev_traded = start_date, 0.0
    for cp in cps:
        valL, valS = ckptL.get(cp, prev_cpL), ckptS.get(cp, prev_cpS)
        traded_to_cp = cumL.get(cp, 0.0) + cumS.get(cp, 0.0)
        elapsed = (cp - prev_cp_date).days
        # 일별 실제 대차잔고 대신 이 구간에서 체결된 명목 숏 비중(achieved_s)을 대차비용 계산 기준으로 사용한다.
        # 연 3% 고정 대차료를 체크포인트 사이 경과일수/365로 단순 비례 배분한다.
        cost = (traded_to_cp - prev_traded) * P['spread'] / 2 + achieved_s * P['borrow_annual'] * elapsed / 365.0
        dV = (valL - prev_cpL) - (valS - prev_cpS) - cost
        rets[cp] = dV / V
        costs[cp] = cost
        long_vals[cp] = valL
        short_vals[cp] = valS
        V += dV
        prev_cpL, prev_cpS, prev_cp_date, prev_traded = valL, valS, cp, traded_to_cp

    return dict(start=start_date, end=end_date, long_ranked=long_ranked, short_ranked=short_ranked,
                long_target=lt, short_target=st, prev_long_w=prev_long_w, prev_short_w=prev_short_w,
                long_contrib=rL, short_contrib=rS, traded_long=traded_l, traded_short=traded_s,
                achieved_long=achieved_l, achieved_short=achieved_s, hold_long=hold_l, hold_short=hold_s,
                next_long_w=next_long_w, next_short_w=next_short_w,
                rets=pd.Series(rets, dtype=float), costs=pd.Series(costs, dtype=float), window_return=V - 1,
                long_vals=pd.Series(long_vals, dtype=float), short_vals=pd.Series(short_vals, dtype=float))


# 포트폴리오 하나를 schedule(리밸런싱 날짜)대로 끝까지 돌린다. 반환: (월별 수익률, 리밸런싱별 회전율, 보유 내역)
def run_strategy_port(schedule, calendar, P, port_name='port_A', log=None):
    """포트폴리오 하나를 schedule(리밸런싱 날짜)대로 끝까지 돌린다. 반환: (월별 수익률, 리밸런싱별 회전율, 보유 내역).
    회전율은 총자산(AUM=1000억) 대비 거래금액이다: traded_long/traded_short는 이 포트폴리오 자금(P['aum']=333.3억) 대비 거래 비중이므로
    P['aum']/AUM을 곱해 전체 1000억 기준으로 환산한다(포트폴리오당 333.3억이면 대략 1/3배)."""
    rets, turnover, records = [], pd.Series(dtype=float), []
    prev_l = prev_s = pd.Series(dtype=float)
    for i in range(len(schedule) - 1):
        s, e = schedule[i], schedule[i + 1]
        W = strategy_window(s, e, prev_l, prev_s, _checkpoints(s, e, calendar), P)
        rets.append(W['rets'])
        turnover.loc[s] = (W['traded_long'] + W['traded_short']) * P['aum'] / AUM
        records += [(s, 'Long', t, w) for t, w in W['hold_long'].items()]
        records += [(s, 'Short', t, w) for t, w in W['hold_short'].items()]
        # 상세 검증용 로그가 요청된 실행에서만 구간별 중간 결과를 저장한다.
        if log is not None:
            W['port'] = port_name
            log.append(W)
        prev_l, prev_s = W['next_long_w'], W['next_short_w']
    return pd.concat(rets), turnover, records


# port_A·B·C를 각자의 리밸런싱 일정으로 독립 실행하고 월별 수익률을 평균한다
def run_strategy(P, schedules, calendar, log=None):
    """port_A·B·C를 각자의 리밸런싱 일정으로 독립 실행하고 월별 수익률을 평균한다.
    운용규모는 1000억 고정(세 포트폴리오 모두 항상 333.3억)이라 세 포트폴리오의 비중은 매달 같다.
    성과는 세 포트폴리오가 모두 투자된 시점(perf_start)부터 잰다. 첫 행은 그 시점(수익률 0, NAV=1)이다."""
    outs = [run_strategy_port(s, calendar, P, PORT_NAMES[k], log) for k, s in enumerate(schedules)]
    t0 = max(s[0] for s in schedules)   # 성과 측정 시작점: 모든 포트폴리오가 첫 편입을 시작한 날(가장 늦은 포트폴리오의 첫 리밸런싱일)
    ret = pd.concat([o[0] for o in outs], axis=1).mean(axis=1)
    ret = pd.concat([pd.Series({t0: 0.0}), ret[ret.index > t0]]).sort_index()
    ret.index.name = None
    turnover = pd.concat([o[1] for o in outs]).sort_index()
    turnover = turnover[turnover.index >= t0]
    records = [r for o in outs for r in o[2]]
    return ret, turnover, records


# 백테스트 파라미터 묶음. 기본값은 본 전략 설정이고, 민감도 분석은 overrides로 하나씩 바꾼다
def strategy_params(**overrides):
    """백테스트 파라미터 묶음. 기본값은 본 전략 설정이고, 민감도 분석은 overrides로 하나씩 바꾼다."""
    P = dict(factor=csi_factor, tau=TAU, n=n, cap_pct=CAP_PCT,
             long_k=LONG_WEIGHT_K, short_k=SHORT_WEIGHT_K, weight_mode='strategy',
             aum=AUM_PORT, participation=PARTICIPATION_RATE, max_days=MAX_EXECUTION_DAYS,
             spread=SPREAD, borrow_annual=BORROW_ANNUAL)
    P.update(overrides)
    return P
```

## **4. 첫 리밸런싱 예시: port_A, 2011-12-31 ~ 2012-03-31**

전체 백테스트와 같은 함수(`strategy_window`)·같은 파라미터로 한 구간만 계산해 3-1 → 3-5의 흐름을 보여준다.
이 구간 수익률은 전체 백테스트의 port_A 첫 구간과 정확히 같다(5장에서 확인). 2012년 1~2월은 성과 측정 시작(2012-02-29) 전이라 헤드라인에는 3월분만 들어간다.

```python
P = strategy_params()
start_date, end_date = port_schedules[0][0], port_schedules[0][1]
cps = _checkpoints(start_date, end_date, all_month_ends)
W0 = strategy_window(start_date, end_date, pd.Series(dtype=float), pd.Series(dtype=float), cps, P)
print(f"구간: {start_date.date()} ~ {end_date.date()}, 월말 평가 시점: {[d.date() for d in cps]}")
```

**3-1·3-2: 종목 선정과 목표비중** (롱: 감소규모 × 점진성 / 숏: 저변동성)

```python
c = csi_factor.loc[start_date]
long_top, short_top = W0['long_ranked'][:n], W0['short_ranked'][:n]
real = [t for t in long_top if c[t] < -REAL_CHANGE_EPS]
lump = _lump_dn(start_date, real, TAU)
score = (-c.reindex(real)) * (1 - lump)
pr = (score.rank(method='average') - 1) / (len(score) - 1)   # 동점이면 평균 순위

long_tbl = pd.DataFrame({'CSI': c.reindex(long_top),
                         '감소한 달 수': (_d_monthly.loc[:start_date].iloc[-TAU:][real] < -MONTH_EPS).sum().reindex(long_top),
                         '가장 큰 한 달 비중': lump.reindex(long_top),
                         '점진성': (1 - lump).reindex(long_top),
                         '점수(감소규모x점진성)': score.reindex(long_top),
                         '순위(0~1)': pr.reindex(long_top),
                         '가중 1+k(2순위-1)': _rank_tilt(score, LONG_WEIGHT_K).reindex(long_top),
                         '목표비중': pd.Series(W0['long_target'])})
long_tbl['동일가중 대비'] = long_tbl['목표비중'] * n
short_tbl = pd.DataFrame({'CSI': c.reindex(short_top),
                          '60거래일 변동성': _vol60_monthly.loc[start_date].reindex(short_top),
                          '목표비중': pd.Series(W0['short_target'])})
short_tbl['동일가중 대비'] = short_tbl['목표비중'] * n

print(f"롱 {len(long_top)}종목 (모두 실제로 주식수가 줄어든 종목), 숏 {len(short_top)}종목")
display(long_tbl.sort_values('목표비중', ascending=False).round(4))
display(short_tbl.sort_values('목표비중', ascending=False).round(4))
```

**3-4: 거래대금 제약 없는 체결** (port_A 333.3억 기준, 가격이 있는 첫 거래일에 목표비중 전량 체결)


```python
fill = {}
for leg, tgt, hold in (('Long', W0['long_target'], W0['hold_long']), ('Short', W0['short_target'], W0['hold_short'])):
    f = pd.DataFrame({'목표비중': pd.Series(tgt), '체결비중': pd.Series(hold, dtype=float).reindex(list(tgt)).fillna(0.0)})
    f['체결률'] = f['체결비중'].clip(upper=f['목표비중']) / f['목표비중']
    fill[leg] = f
    print(f"{leg}: 목표 100% 중 {f['체결비중'].sum():.1%} 체결, 목표를 다 채운 종목 {(f['체결률'] >= 0.999999).sum()}/{len(f)}개, 현금 {1 - f['체결비중'].sum():.1%}")
pd.concat(fill).round(4)
```

**3-5: 구간 수익률** (롱 − 숏 − 거래비용, 월별 수익률을 복리로 곱하면 구간 수익률)

```python
print('롱 레그 기여수익률 :', round(W0['long_contrib'], 4))
print('숏 레그 기여수익률 :', round(W0['short_contrib'], 4))
print('거래비용 합계      :', round(W0['costs'].sum(), 4))
print('구간 수익률        :', round(W0['window_return'], 4))
print()
print('월별 수익률:')
print(W0['rets'].round(4).to_string())
```

```python
# 벤치마크: KOSPI 월간 수익률 (월 마지막 영업일 → 달력 월말로 맞춤)
factor_raw = pd.read_csv("Performance Evalution input/factor_monthly.csv", index_col=0, parse_dates=True)
kospi_m = factor_raw['코스피'].astype(float).pct_change(fill_method=None).dropna()
kospi_m.index = kospi_m.index + pd.offsets.MonthEnd(0)
kospi_m = kospi_m.loc[df.index[0]:df.index[-1]]

# 무위험수익률: factor_monthly.csv의 RF(연율 %) -> 월 수익률(RF/100/12), 달력 월말로 맞춤. Sharpe는 초과수익(수익률 - RF) 기준
rf_m = factor_raw['RF'].astype(float) / 100 / 12
rf_m.index = rf_m.index + pd.offsets.MonthEnd(0)
rf_m = rf_m[~rf_m.index.duplicated(keep='last')]


# 첫 행 = 시작점(수익률 0, NAV=1). MDD는 시작점을 포함한 월말 NAV 기준
def performance_metrics(portfolio, periods_per_year=12):
    """첫 행 = 시작점(수익률 0, NAV=1). MDD는 시작점을 포함한 월말 NAV 기준."""
    rets = portfolio['Return'].iloc[1:]
    nav = portfolio['NAV'] / portfolio['NAV'].iloc[0]
    years = len(rets) / periods_per_year
    vol = rets.std() * np.sqrt(periods_per_year)
    excess = rets - rf_m.reindex(rets.index).ffill()   # 월별 초과수익률 (무위험수익률 = factor_monthly.csv의 RF)
    return {"CAGR": nav.iloc[-1] ** (1 / years) - 1, "Volatility (ann.)": vol,
            "Sharpe Ratio": excess.mean() / excess.std() * np.sqrt(periods_per_year), "MDD": (nav / nav.cummax() - 1).min(),
            "Average Turnover": portfolio['Turnover'].iloc[1:].mean()}


metrics = performance_metrics(df)
IS_END = pd.Timestamp('2020-12-31')
t0, t1 = df.index[0], df.index[-1]
colors = {"CSI Strategy": "#2a78d6", "KOSPI": "#eb6834"}


# (start, end] 구간의 월별 수익률 앞에 start 시점(수익률 0)을 붙여 NAV=1에서 다시 시작하게 한다
def cut(ret, start, end):
    """(start, end] 구간의 월별 수익률 앞에 start 시점(수익률 0)을 붙여 NAV=1에서 다시 시작하게 한다."""
    s = ret[(ret.index > start) & (ret.index <= end)].dropna()
    return pd.concat([pd.Series({start: 0.0}), s])


# (start, end] 구간의 CSI 전략과 KOSPI 성과를 나란히 (둘 다 start 시점 NAV=1에서 출발)
def perf_vs_kospi(start, end):
    """(start, end] 구간의 CSI 전략과 KOSPI 성과를 나란히 (둘 다 start 시점 NAV=1에서 출발)."""
    out = {}
    for name, r in [("CSI Strategy", df['Return']), ("KOSPI", kospi_m)]:
        seg = cut(r, start, end)
        turnover = df['Turnover'].reindex(seg.index) if name == "CSI Strategy" else np.nan
        out[name] = performance_metrics(pd.DataFrame({'Return': seg, 'NAV': (1 + seg).cumprod(), 'Turnover': turnover}))
    return pd.DataFrame(out).round(4)


# 위: CSI 전략·KOSPI NAV / 아래: 낙폭. 제목에 CSI 전략의 CAGR·Sharpe·MDD
def nav_dd_panels(axes_col, start, end, title):
    """위: CSI 전략·KOSPI NAV / 아래: 낙폭. 제목에 CSI 전략의 CAGR·Sharpe·MDD."""
    for name, r in [("CSI Strategy", df['Return']), ("KOSPI", kospi_m)]:
        nav = (1 + cut(r, start, end)).cumprod()
        axes_col[0].plot(nav.index, nav.values, color=colors[name], linewidth=2, label=name)
        axes_col[0].annotate(f"{nav.iloc[-1]:.2f}", (nav.index[-1], nav.iloc[-1]), xytext=(4, 0), textcoords="offset points",
                             va="center", fontsize=9, color="#333333")
        axes_col[1].plot(nav.index, (nav / nav.cummax() - 1).values, color=colors[name], linewidth=1.5)
    seg = cut(df['Return'], start, end)
    m = performance_metrics(pd.DataFrame({'Return': seg, 'NAV': (1 + seg).cumprod(), 'Turnover': 0.0}))
    axes_col[0].set_title(f"{title}\nCSI: CAGR {m['CAGR']:.1%} | Sharpe {m['Sharpe Ratio']:.2f} | MDD {m['MDD']:.1%}", fontsize=11)
    axes_col[0].axhline(1.0, color="#000000", alpha=0.15, linewidth=1)
    axes_col[1].axhline(0.0, color="#000000", alpha=0.15, linewidth=1)
    axes_col[1].yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
    for ax in axes_col:
        ax.grid(axis="y", color="#000000", alpha=0.08)
        for sp in ("top", "right"):
            ax.spines[sp].set_visible(False)
```

### **6-5. 롱 / 숏 레그별 기여**

롱 레그만 운용했을 때, 숏 레그만 운용했을 때의 NAV를 각각 구해 표본내·표본외로 나눠 비교한다(비용은 두 레그에 공통으로 들어가 있어 레그별로 나누지 않고, 순수 가격 수익만 비교한다).
전체 기간을 한 그래프에 이어 그리면 최근 롱의 누적 배수가 훨씬 커서 초반 숏의 기여가 평평해 보이는 착시가 생기므로, 두 구간을 각각 NAV=1부터 다시 그린다.
6-4에서 본 것처럼 주식수를 줄이는 기업이 늘면서, 표본외에서는 롱 레그의 기여가 크게 커졌다.

```python
# 레그 단독 월별 수익률(비용 제외, gross): 포트폴리오(port_A/B/C)마다 구간별 체크포인트 값을 이어 붙여 수익률로 바꾸고,
def leg_return_series(value_key, invert=False):
    """레그 단독 월별 수익률(비용 제외, gross): 포트폴리오(port_A/B/C)마다 구간별 체크포인트 값을 이어 붙여 수익률로 바꾸고,
    전체 수익률과 같이 포트폴리오 평균을 낸다. 숏은 기초가치가 오르면 손실이므로 invert=True로 부호를 뒤집는다."""
    port_rets = {}
    for W in backtest_log:
        prev = 1.0
        for cp, val in W[value_key].items():
            r = val / prev - 1
            port_rets.setdefault(W['port'], {})[cp] = -r if invert else r
            prev = val
    combined = pd.DataFrame({p: pd.Series(v) for p, v in port_rets.items()}).sort_index().mean(axis=1)
    return pd.concat([pd.Series({t0: 0.0}), combined[combined.index > t0]]).sort_index()


long_ret = leg_return_series('long_vals')
short_ret = leg_return_series('short_vals', invert=True)


# 지정 구간의 레그 단독 월수익률에서 CAGR, Sharpe, MDD를 계산한다
def leg_perf(ret, start, end):
    """지정 구간의 레그 단독 월수익률에서 CAGR, Sharpe, MDD를 계산한다."""
    seg = cut(ret, start, end)
    nav = (1 + seg).cumprod()
    r = seg.iloc[1:]
    years = len(r) / 12
    return {'CAGR': nav.iloc[-1] ** (1 / years) - 1, 'Sharpe': (r - rf_m.reindex(r.index).ffill()).mean() / (r - rf_m.reindex(r.index).ffill()).std() * np.sqrt(12), 'MDD': (nav / nav.cummax() - 1).min()}


leg_table = pd.DataFrame({('표본내', 'Long leg'): leg_perf(long_ret, t0, IS_END), ('표본내', 'Short leg'): leg_perf(short_ret, t0, IS_END),
                          ('표본외', 'Long leg'): leg_perf(long_ret, IS_END, t1), ('표본외', 'Short leg'): leg_perf(short_ret, IS_END, t1)}).T
leg_table.round(4)
```

```python
leg_series = {'Long leg': long_ret, 'Short leg': short_ret, 'Combined (실제)': df['Return'], 'KOSPI': kospi_m}
leg_colors = {'Long leg': "#2a78d6", 'Short leg': "#eb6834", 'Combined (실제)': "#7b4fc9", 'KOSPI': "#333333"}


# 롱·숏·결합전략·KOSPI의 구간 NAV와 낙폭을 두 패널에 그린다
def leg_nav_panel(ax_top, ax_dd, start, end, title):
    """롱·숏·결합전략·KOSPI의 구간 NAV와 낙폭을 두 패널에 그린다."""
    for name, r in leg_series.items():
        nav = (1 + cut(r, start, end)).cumprod()
        lw, ls = {'Combined (실제)': (2.0, '--'), 'KOSPI': (1.6, ':')}.get(name, (1.6, '-'))
        ax_top.plot(nav.index, nav.values, color=leg_colors[name], linewidth=lw, linestyle=ls, label=name)
        ax_top.annotate(f"{nav.iloc[-1]:.2f}", (nav.index[-1], nav.iloc[-1]), xytext=(4, 0), textcoords="offset points",
                        va="center", fontsize=9, color="#333333")
        ax_dd.plot(nav.index, (nav / nav.cummax() - 1).values, color=leg_colors[name], linewidth=1.3, linestyle=ls)
    lp, sp_ = leg_perf(long_ret, start, end), leg_perf(short_ret, start, end)
    ax_top.set_title(f"{title}\nLong: CAGR {lp['CAGR']:.1%} Sharpe {lp['Sharpe']:.2f} | Short: CAGR {sp_['CAGR']:.1%} Sharpe {sp_['Sharpe']:.2f}", fontsize=10)
    ax_top.axhline(1.0, color="#000000", alpha=0.15, linewidth=1)
    ax_dd.axhline(0.0, color="#000000", alpha=0.15, linewidth=1)
    ax_dd.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
    for ax in (ax_top, ax_dd):
        ax.grid(axis="y", color="#000000", alpha=0.08)
        for spine in ("top", "right"):
            ax.spines[spine].set_visible(False)


fig, axes = plt.subplots(2, 2, figsize=(13, 7), sharex='col', gridspec_kw={'height_ratios': [2.2, 1]})
leg_nav_panel(axes[0, 0], axes[1, 0], t0, IS_END, "In-sample (2012.03-2020.12)")
leg_nav_panel(axes[0, 1], axes[1, 1], IS_END, t1, "Out-of-sample (2021.01-2026.08)")
axes[0, 0].set_ylabel("NAV (rebased to 1, 비용 제외)")
axes[1, 0].set_ylabel("Drawdown")
axes[0, 0].legend(frameon=False, loc="upper left")
plt.tight_layout()
plt.show()
```

비중은 각 레그의 포트폴리오 자금(333.3억)을 1로 본 값이다. 상태: `new`(신규 편입) / `kept`(유지) / `unfilled`(가격 부재로 목표에 못 미친 채 편입 종료) / `exit_pending`(빠졌지만 한도 때문에 못 팔고 남음) / `exit_done`(전량 매도).

```python
_cal = adj_close.index
detail_rows, summary_rows = [], []
for W in backtest_log:
    s = W['start']
    adv_row, vol_row = trading_value_20.loc[_cal[_cal <= s][-1]], _vol60_monthly.loc[s]
    csi_row, cap_row = csi_factor.loc[s], market_cap_monthly.loc[s]
    for leg in ('Long', 'Short'):
        lk = leg.lower()
        ranked, tgt = W[f'{lk}_ranked'], W[f'{lk}_target']
        prev, hold, nxt = W[f'prev_{lk}_w'], W[f'hold_{lk}'], W[f'next_{lk}_w']
        top_n = set(ranked[:n])
        ranks = {t: k + 1 for k, t in enumerate(ranked)}
        for t in sorted(set(prev.index) | set(hold) | top_n | set(nxt.index), key=lambda x: (ranks.get(x, 10**6), str(x))):
            p, h, e = float(prev.get(t, 0.0)), float(hold.get(t, 0.0)), float(nxt.get(t, 0.0))
            tw = tgt.get(t, 0.0) if t in top_n else 0.0
            # 현재 목표 후보인지에 따라 편입 상태와 청산 상태를 구분해 기록한다.
            if t in top_n:
                status = ('new' if p <= 1e-9 else 'kept') + ('/unfilled' if h < tw - 1e-6 else '')
            else:
                status = 'exit_pending' if h > 1e-9 else 'exit_done'
            adv = adv_row.get(t, np.nan)
            detail_rows.append(dict(
                Port=W['port'], Date=s, EndDate=W['end'], Leg=leg, Ticker=t, Rank=ranks.get(t, np.nan),
                CSI=csi_row.get(t, np.nan), MarketCap=cap_row.get(t, np.nan),
                Vol60=vol_row.get(t, np.nan), ADV20=adv,
                DailyCap_W=np.inf,  # 거래대금 제약 없음
                Target_W=tw, Prev_W=p, PostExec_W=h, WindowEnd_W=e, Status=status))
        summary_rows.append(dict(
            Port=W['port'], Date=s, EndDate=W['end'], Leg=leg,
            Contrib=W[f'{lk}_contrib'], Traded_W=W[f'traded_{lk}'], N_Target=len(tgt),
            N_Hold=sum(1 for w in hold.values() if w > 1e-9), Invested_W=sum(hold.values()),
            Cash_W=max(1.0 - sum(hold.values()), 0.0),
            N_New=sum(1 for t in top_n if prev.get(t, 0.0) <= 1e-9),
            N_ExitPending=sum(1 for t, w in hold.items() if t not in top_n and w > 1e-9),
            Window_Return=W['window_return']))

detail, summary = pd.DataFrame(detail_rows), pd.DataFrame(summary_rows)

os.makedirs("output", exist_ok=True)
detail.to_csv("output/holdings_detail.csv", index=False, encoding="utf-8-sig")
summary.to_csv("output/quarter_summary.csv", index=False, encoding="utf-8-sig")
print(f"holdings_detail {detail.shape} / quarter_summary {summary.shape}")
detail[detail["Date"] == detail["Date"].max()].head(12)
```

```python
# 이 셀에서만 거래대금 제약을 적용한 체결률 검증
# 조건: 종목당 하루 최대 20일 평균 거래대금의 10%, 편입 창 최대 10영업일
# 새 커널에서 이 셀만 실행한 경우, 필요한 앞쪽 계산 셀만 자동 실행한다.
_required = ('run_strategy', 'strategy_params', 'port_schedules', 'all_month_ends')
# 핵심 함수나 일정 변수가 하나라도 없으면 노트북 초기화가 필요한 상태로 판단한다.
if not all(name in globals() for name in _required):
    import json as _json
    from pathlib import Path as _Path
    import os as _os

    # 현재 폴더를 우선 확인하고, 없으면 하위 폴더에서 현재 노트북 파일을 찾는다.
    _cwd = _Path.cwd()
    _nb_candidates = list(_cwd.glob('CSI_strategy.ipynb')) or list(_cwd.glob('**/CSI_strategy.ipynb'))
    # 노트북 파일을 찾지 못하면 필요한 앞쪽 셀을 불러올 수 없으므로 즉시 중단한다.
    if not _nb_candidates:
        raise FileNotFoundError('CSI_strategy.ipynb를 현재 폴더 또는 하위 폴더에서 찾을 수 없습니다.')
    _nb_path = _nb_candidates[0]
    _nb = _json.loads(_nb_path.read_text(encoding='utf-8'))
    # 데이터 로딩, 팩터·가중치, 일정, 체결 엔진 정의에 필요한 코드 셀만 선택한다.
    _setup_cells = [6, 7, 9, 11, 13, 15, 17, 19, 21]
    print('새 커널 감지: 체결률 계산에 필요한 앞쪽 셀을 자동 초기화합니다.')
    try:
        _os.chdir(_nb_path.parent)  # 입력 파일의 상대경로 기준을 노트북 폴더로 맞춤
        # 선택한 셀을 원래 순서대로 실행해 체결률 계산에 필요한 상태를 구성한다.
        for _cell_no in _setup_cells:
            _src = ''.join(_nb['cells'][_cell_no].get('source', []))
            exec(compile(_src, f'{_nb_path.name}:cell_{_cell_no}', 'exec'), globals())
    finally:
        # 초기화 성공 여부와 관계없이 사용자가 실행하던 원래 작업 폴더로 돌아간다.
        _os.chdir(_cwd)

# 기존 무제약 백테스트와 분리된 로그에 ADV20 10% 제약 시뮬레이션 결과를 저장한다.
_liq_fill_log = []
_ = run_strategy(
    strategy_params(participation=0.10, max_days=10),
    port_schedules, all_month_ends, log=_liq_fill_log
)

# 각 포트폴리오·리밸런싱일·레그별 목표량과 실제 체결량을 집계한다.
_liq_fill_rows = []
for W in _liq_fill_log:
    for leg in ('Long', 'Short'):
        lk = leg.lower()
        tgt = pd.Series(W[f'{lk}_target'], dtype=float)
        hold = pd.Series(W[f'hold_{lk}'], dtype=float)
        # 목표 초과 보유분은 체결률에 포함하지 않고, 목표 종목의 미체결분은 0으로 채운다.
        got = hold.reindex(tgt.index).fillna(0.0).clip(upper=tgt)
        _liq_fill_rows.append(dict(
            Port=W['port'], Date=W['start'], Leg=leg,
            TargetAmount=tgt.sum(), FilledAmount=got.sum(),
            TargetNames=len(tgt), FilledNames=(got >= tgt * (1 - 1e-6)).sum()))

# 롱·숏 원자료를 만들고 같은 포트폴리오·리밸런싱일 기준 합산 행을 추가한다.
_liq_fill_base = pd.DataFrame(_liq_fill_rows)
_liq_fill_combined = (_liq_fill_base.groupby(['Port', 'Date'], as_index=False)
                      [['TargetAmount', 'FilledAmount', 'TargetNames', 'FilledNames']].sum()
                      .assign(Leg='Long+Short'))
_liq_fill = pd.concat([_liq_fill_base, _liq_fill_combined], ignore_index=True)
# 금액 체결률은 목표금액 대비 체결금액, 종목 체결률은 목표를 전량 채운 종목의 비율이다.
_liq_fill['금액체결률'] = _liq_fill['FilledAmount'] / _liq_fill['TargetAmount']
_liq_fill['종목체결률'] = _liq_fill['FilledNames'] / _liq_fill['TargetNames']

# Long, Short, Long+Short별 평균 체결률과 가장 낮았던 리밸런싱의 체결률을 요약한다.
_liq_fill_summary = (_liq_fill.groupby('Leg')
                     .agg(평균_금액체결률=('금액체결률', 'mean'),
                          최소_금액체결률=('금액체결률', 'min'),
                          평균_종목체결률=('종목체결률', 'mean'),
                          최소_종목체결률=('종목체결률', 'min'))
                     .reindex(['Long', 'Short', 'Long+Short']))
print('거래대금 제약: 일별 ADV20의 10%, 편입 최대 10영업일')
print(f"리밸런싱 {len(_liq_fill_base) // 2}회(port_A·B·C 합계), {_liq_fill_base['Date'].min().date()} ~ {_liq_fill_base['Date'].max().date()}")
display(_liq_fill_summary.style.format('{:.1%}'))
```
